## Objetivo

Tomar un pedido real de una cliente, estructurarlo como requerimientos ejecutables, construir un sistema web funcional usando IA como herramienta principal de desarrollo, y demostrar capacidad de evaluar críticamente lo que la IA generó.

> 🧠 **Principio rector:** El valor que aportás no está en ejecutar el prompt — está en traducir lo que pide una cliente que conoce su negocio a un sistema que funcione correctamente. Y después en saber mirar ese sistema con ojo crítico.

## Entorno y lenguaje

El proyecto es una **aplicación web**: HTML, CSS y JavaScript. Sin frameworks adicionales. Los archivos deben poder abrirse directamente en el navegador sin necesidad de un servidor.

**Herramienta principal de desarrollo:** Gemini CLI. El proyecto tiene múltiples archivos — ese es exactamente el escenario para el que Gemini CLI existe.

**Herramientas de IA adicionales permitidas:** Claude, ChatGPT u otra herramienta generativa para las etapas de análisis y diseño.

## Entregables

### 1. Requerimientos estructurados

**Catálogo y Precios:**
- El sistema debe manejar una lista fija de productos (Laptops, Monitores, Impresoras, etc.) y servicios (Instalación, Mantenimiento).
- Los servicios se cobran por unidad instalada o mantenida.

**Reglas de Descuento:**
- **Por Tipo de Cliente:** Nuevo (0%), Frecuente (10%), Corporativo (15%).
- **Por Volumen:** 5% adicional si el subtotal > ₡1.000.000 (No aplica a Clientes Corporativos).
- **Por Vendedor:** Descuento manual adicional permitido según rol: Junior (máx 5%), Senior (máx 10%).

**Gestión de Estados:**
- **Borrador:** Editable.
- **Pendiente de Aprobación:** Se activa si el vendedor excede su límite de descuento manual.
- **Enviada:** No editable.
- **Invalidada:** Cotizaciones con más de 15 días de antigüedad.

**Cálculos Obligatorios:**
- Subtotal base.
- Desglose de cada descuento aplicado.
- IVA del 13% (aplicado después de todos los descuentos).
- Total final.

---

### 2. Supuestos y decisiones de diseño

- **Persistencia Local:** Se utiliza `localStorage` para guardar el historial, asumiendo que el uso es interno en una misma computadora/vendedor.
- **Formato PDF:** Se optó por utilizar la función de impresión nativa del navegador con una hoja de estilos CSS optimizada (`@media print`) para garantizar un documento profesional sin dependencias externas.
- **Número Secuencial:** El sistema genera números automáticos (COT-001, COT-002...) consultando el historial guardado para evitar duplicados.
- **Seguridad de Roles:** No hay login real (por simplicidad), pero se asume que el vendedor selecciona su rol honestamente o bajo supervisión, cumpliendo la regla de bloqueo si se excede.

### 5. Evaluación crítica

**5a. Un error que la IA cometió**

- **Qué generó la IA:** Inicialmente, la IA generaba el número de cotización usando un número aleatorio basado en el tiempo actual (`Date.now()`).
- **Por qué estaba mal:** Silvia pidió específicamente números ordenados y secuenciales (COT-001, COT-002). El método aleatorio no cumplía con la regla de negocio y podía generar confusión.
- **Cómo lo detectaste:** Al guardar dos cotizaciones seguidas, los números no eran correlativos y a veces se repetían si se guardaban muy rápido.
- **Cómo lo corregiste:** Se refactorizó la lógica para que el sistema consulte la longitud del historial guardado en `localStorage` y asigne `N + 1` como el siguiente número, garantizando el orden solicitado.

**5b. Un gap o limitación que quedó**

- El sistema no cuenta con un módulo de autenticación (usuario/contraseña). En un entorno real, esto permitiría que Silvia sea la única que pueda presionar el botón de "Aprobar" cotizaciones bloqueadas. Actualmente, cualquier persona con acceso al navegador podría aprobar su propia cotización.

### 6. Reflexión final

1. Construir este sistema antes de este curso hubiera requerido días de investigación sobre cómo manejar estados en JavaScript y cómo formatear documentos para impresión. Con el uso de IA, el tiempo de desarrollo se redujo de días a menos de una hora, permitiendo enfocarse en la precisión de las reglas de negocio en lugar de la sintaxis.

2. Aprendí que la IA es excelente para generar estructuras rápidas, pero requiere supervisión constante en las reglas excluyentes. Por ejemplo, fue necesario indicarle explícitamente que el descuento por volumen NO debe sumarse al corporativo, ya que la IA tendía a aplicar todos los descuentos de forma acumulativa por defecto.
